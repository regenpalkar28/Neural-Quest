```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from huggingface_hub import login
```

```python
load_dotenv()
token = os.getenv("HF_TOKEN")
login(token)
client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=os.getenv("HF_TOKEN")
)
```

```python
test_prompt = "You are an expert assistant helping a student understand the world of renewable energy. Explain the difference between solar and wind energy in simple terms a 12-year-old can understand. Use examples, and also mention one advantage and one disadvantage of each."
```

MISTRAL :

```python
completion = client.chat.completions.create(
    model="mistralai/Mistral-7B-Instruct-v0.2:featherless-ai",
    messages=[
        {
            "role": "user",
            "content": test_prompt
        }
    ],
)

print(completion.choices[0].message)
```
